### The 3 fuel types in this dataset

Running `df['fuel_type'].unique()` shows the dataset contains **Gasoline**, **Diesel**, and **Hybrid** vehicles:

- **Gasoline** (6,913 records) — The most common fuel type. Gasoline engines use spark-ignition combustion, where an air-fuel mixture is ignited by a spark plug. They are widely used because gasoline is energy-dense, easy to distribute, and the engines are relatively lightweight and inexpensive to produce, though they are generally less fuel-efficient than diesel or hybrid alternatives.
- **Diesel** (2,053 records) — Diesel engines use compression-ignition combustion, where fuel is injected into highly compressed air and ignites from heat alone (no spark plug). Diesel fuel has a higher energy density than gasoline, giving diesel engines better fuel economy and more torque, which is why they are popular in trucks and long-distance vehicles. They also tend to have higher NOx and particulate emissions.
- **Hybrid** (1,034 records) — Hybrid vehicles combine an internal combustion engine (usually gasoline) with one or more electric motors and a battery pack. The electric motor assists during acceleration and low-speed driving, and can capture energy through regenerative braking, which significantly improves overall fuel efficiency compared to conventional gasoline or diesel vehicles.

## Q6. Median value of horsepower
Find the median value of the horsepower column in the dataset.
Next, calculate the most frequent value of the same horsepower column.
Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
Now, calculate the median value of horsepower once again.
Has it changed?

```python
median_horsepower_before = df['horsepower'].median()
most_frequent_horsepower = df['horsepower'].mode()[0]
print("Median before:", median_horsepower_before)
print("Most frequent:", most_frequent_horsepower)
```

```python
horsepower_filled = df['horsepower'].fillna(most_frequent_horsepower)
median_horsepower_after = horsepower_filled.median()
print("Median after:", median_horsepower_after)
if median_horsepower_after > median_horsepower_before:
    answer_q6 = "Yes, it increased"
elif median_horsepower_after < median_horsepower_before:
    answer_q6 = "Yes, it decreased"
else:
    answer_q6 = "No"
answer_q6
```

## Q7. Sum of weights
- Select all the cars from Asia
- Select only columns vehicle_weight and model_year
- Select the first 7 values
- Get the underlying NumPy array (X)
- Compute XTX = X.T @ X
- Invert XTX
- Create y = [1100, 1300, 800, 900, 1000, 1100, 1200]
- w = inv(XTX) @ X.T @ y
- Sum of all elements of w?

```python
import numpy as np

X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).to_numpy()

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y
sum_w = w.sum()
sum_w
```
